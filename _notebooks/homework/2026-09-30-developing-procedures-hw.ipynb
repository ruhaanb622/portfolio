## Practice 1: Rename Procedures Clearly

Descriptive procedure names make the call sequence readable without opening every function.

````python
# CODE_RUNNER: 3.13 Practice 1 - descriptive procedure names

# Each helper name describes exactly one dessert-making responsibility.
def measure_flour():
    print("Measuring 2 cups of flour.")

def mix_dessert_ingredients():
    print("Mixing flour, sugar, and eggs.")

def bake_dessert():
    print("Baking at 350°F for 25 minutes.")

# The main procedure reads like a short plan because the helper names explain the steps.
def make_dessert():
    measure_flour()
    mix_dessert_ingredients()
    bake_dessert()

make_dessert()
````

## Practice 2: Equivalent Java Methods

```java
public static void makeDessert() {
    measureFlour();
    mixDessertIngredients();
    bakeDessert();
}

private static void measureFlour() {
    System.out.println("Measuring 2 cups of flour.");
}

private static void mixDessertIngredients() {
    System.out.println("Mixing flour, sugar, and eggs.");
}

private static void bakeDessert() {
    System.out.println("Baking at 350°F for 25 minutes.");
}
```

`public` makes `makeDessert()` callable from outside its class. The helper methods can be `private` because only the class itself needs to call them.

## Practice 3: Split a Long Procedure

Breaking one large procedure into small helpers makes each step easier to test independently.

````python
# CODE_RUNNER: 3.13 Practice 3 - decompose a long procedure

# Each helper returns a value so the step can be tested without running the entire program.
def choose_ingredient():
    return "flour"

def mix(item):
    return "Mixing " + item + " into the bowl."

def heat():
    return "Heating the oven to 350°F."

def finish():
    return "Dessert is ready!"

# The outer procedure combines the small, testable pieces into the full workflow.
def prepare_dessert():
    ingredient = choose_ingredient()
    return [mix(ingredient), heat(), finish()]

for step in prepare_dessert():
    print(step)

# Direct tests make it obvious which helper fails if behavior changes.
print("mix test:", mix("sugar") == "Mixing sugar into the bowl.")
print("heat test:", heat() == "Heating the oven to 350°F.")
````

## Mini-Game Practice: Procedure Naming

The lesson's original mini-game uses `input()`. This runner-friendly version uses a stored name so it can run on the portfolio without waiting for keyboard input.

````python
# CODE_RUNNER: 3.13 Mini-game - show why clear procedure names matter

import random

# These helpers model the same dessert sequence from the lesson mini-game.
def choose_ingredient():
    return random.choice(["flour", "sugar", "eggs", "milk"])

def prepare_ingredient():
    ingredient = choose_ingredient()
    print("Mixing", ingredient, "into the bowl.")
    print("Heating the oven to a friendly temperature...")
    print("Dessert is ready!")

# A descriptive call explains the goal immediately; a vague name like do_step() would not.
print("Readable procedure call: prepare_ingredient()")
prepare_ingredient()
````
