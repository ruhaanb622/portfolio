## SASS Buttons

Completed all three Popcorn Hacks and both Homework Hacks using only OCS SASS button classes and layout containers.

## Popcorn Hack 1: Add the Base Button Class

The base `ocs__btn` class applies the shared OCS button styling without custom CSS.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 1 — Base Button Class | interactive: true -->
<!-- ocs__btn supplies the standard OCS button appearance. -->
<!-- Using an anchor keeps this action semantic because it navigates to a resource page. -->
<a class="ocs__btn" href="https://www.mygoodbrain.org/resources" target="_blank" rel="noopener">Mental Health Resource Directory</a>
```

## Popcorn Hack 2: Modifiers and Button Groups

The first action uses `pill` and `fill`, and both actions are grouped inside an `ocs__links` container.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 2 — Modifiers and Button Group | interactive: true -->
<!-- ocs__links keeps related actions aligned and spaced using the shared SASS system. -->
<!-- pill and fill give the main check-in action stronger visual emphasis. -->
<div class="ocs__links">
  <button class="ocs__btn accent pill fill">Complete Daily Check-In</button>
  <a class="ocs__btn outline" href="https://www.mygoodbrain.org/resources" target="_blank" rel="noopener">Save Draft Journal</a>
</div>
```

## Popcorn Hack 3: Status Tones and Wide Layout Bar

The second action uses `alert-green`, the container is upgraded to `ocs__links--wide`, and the crisis action uses `fill` for maximum priority.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 3 — Status Tones and Wide Layout | interactive: true -->
<!-- ocs__links--wide gives the action bar a wider responsive layout. -->
<!-- Status tone classes communicate meaning without hardcoded colors or custom CSS. -->
<div class="ocs__links--wide">
  <button class="ocs__btn alert-green outline">Complete Check-In</button>
  <a class="ocs__btn alert-red fill" href="https://www.mygoodbrain.org/resources" target="_blank" rel="noopener">Get Crisis Support</a>
</div>
```

## Homework Hack 1: Adaptive Resource Toolbar

The three resource categories are converted from inline styling to the exact OCS utility classes required by the lesson.

```python
%%html
<!-- UI_RUNNER: Homework Hack 1 — Adaptive Resource Toolbar | interactive: true -->
<!-- All three actions use only OCS SASS utility classes; there are no inline style attributes. -->
<!-- Red signals urgent support, green signals self-care, and accent/outline gives community a lower visual priority. -->
<div class="ocs__links">
  <a class="ocs__btn alert-red fill pill" href="https://www.mygoodbrain.org/resources" target="_blank" rel="noopener">Urgent Support</a>
  <a class="ocs__btn alert-green fill pill" href="https://www.mygoodbrain.org/resources" target="_blank" rel="noopener">Self-Care</a>
  <a class="ocs__btn accent outline pill" href="https://www.mygoodbrain.org/resources" target="_blank" rel="noopener">Community</a>
</div>
```

## Homework Hack 2: Primary vs. Secondary Action Priority

The action row uses fill, outline, and alert modifiers to create a clear visual hierarchy without any custom CSS.

```python
%%html
<!-- UI_RUNNER: Homework Hack 2 — Primary vs Secondary Action Priority | interactive: true -->
<!-- The filled accent button is the primary action, while outline buttons remain visually secondary. -->
<!-- The alert-red outline keeps the support link noticeable without competing with the main mood-entry action. -->
<div class="ocs__links">
  <button class="ocs__btn accent fill">Log Mood Entry</button>
  <a class="ocs__btn outline" href="https://www.mygoodbrain.org/resources" target="_blank" rel="noopener">View Weekly Trends</a>
  <a class="ocs__btn alert-red outline" href="https://www.mygoodbrain.org/resources" target="_blank" rel="noopener">Crisis Hotline</a>
</div>
```
