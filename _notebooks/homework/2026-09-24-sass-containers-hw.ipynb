## OCS Container Grammar

The finished [OCS Containers lesson](https://pages.opencodingsociety.com/navigation/sass/containers/lesson) hacks are below. Click **Run** on each labeled UI runner to preview the layout. Both solutions use the shared OCS container grammar with no custom CSS.

## Popcorn Hack: Parade Invite Rescue

The broken parade invite is rebuilt inside one `ocs__container` and one `ocs__card`. The schedule uses a responsive 3-column standard grid, and the start time is accented.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack — Parade Invite Rescue | interactive: true -->
<div class="ocs__container">
  <div class="ocs__card">
    <h2 class="ocs__section-title">Poway Rotary Parade Invite</h2>
    <p class="ocs__description">Join the Poway Scripps Rotary Club for a morning of community, service, and parade fun.</p>
    <div class="ocs__grid ocs__grid--standard cols-3">
      <div class="ocs__grid-cell">Location: Poway Road</div>
      <div class="ocs__grid-cell ocs__grid-cell--accent">Start Time: 9:00 AM</div>
      <div class="ocs__grid-cell">What You Get: Community fun and Rotary spirit</div>
    </div>
  </div>
</div>
```

## Homework Hack: Parade Hub Page

The full parade hub uses exactly one container and one card. It includes the required standard `cols-3` grid with header, accent, and muted cells, an OCS table with two body rows, and a final callout.

```python
%%html
<!-- UI_RUNNER: Homework Hack — Poway Rotary Parade Hub | interactive: true -->
<div class="ocs__container">
  <div class="ocs__card">
    <h2 class="ocs__section-title">Poway Scripps Rotary Parade Hub</h2>
    <p class="ocs__description">Everything volunteers and visitors need for parade day in one responsive Rotary hub.</p>

    <div class="ocs__grid ocs__grid--standard cols-3">
      <div class="ocs__grid-cell ocs__grid-cell--header">Parade Day Information</div>
      <div class="ocs__grid-cell ocs__grid-cell--accent">Parade Start: 9:00 AM on Poway Road</div>
      <div class="ocs__grid-cell ocs__grid-cell--muted">Volunteer Check-In: 8:00 AM</div>
    </div>

    <div class="ocs__table-wrap">
      <table class="ocs__table">
        <tr><th>Event</th><th>Status</th></tr>
        <tr><td>Parade Floats</td><td>Confirmed</td></tr>
        <tr><td>Food Drive</td><td>Collecting donations</td></tr>
      </table>
    </div>

    <div class="ocs__callout">One OCS container keeps the entire Parade hub consistent and responsive on phones and computers.</div>
  </div>
</div>
```
