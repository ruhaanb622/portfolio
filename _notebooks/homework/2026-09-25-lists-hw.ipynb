## Popcorn Hack 2: Sum and Average

```python
# CODE_RUNNER: Lists Popcorn 2
# Add each number with an accumulator instead of using sum().
# Check for an empty list before dividing so the average calculation is safe.

numbers = [7, 6, 8, 5, 7, 9]

total = 0
for number in numbers:
    total += number

average = total / len(numbers)

print("Total:", total)
print("Average:", average)

sleep = []
if len(sleep) == 0:
    print("No data yet")
else:
    sleep_total = 0
    for hours in sleep:
        sleep_total += hours
    print("Sleep average:", sleep_total / len(sleep))
```

## Popcorn Hack 3: Count Then Filter

```python
# CODE_RUNNER: Lists Popcorn 3 - count then filter
# Count each season in a dictionary as it appears in the list.
# Build a second list containing only seasons that appear at least twice.

seasons = ["winter", "summer", "fall", "summer", "spring", "summer", "fall"]

frequency = {}
for season in seasons:
    if season in frequency:
        frequency[season] += 1
    else:
        frequency[season] = 1

popular = []
for season in frequency:
    if frequency[season] >= 2:
        popular.append(season)

print(frequency)
print(popular)
```

## Homework Hack: Indexing, Accumulator, Count, and Filter

```python
# CODE_RUNNER: Lists Homework - indexing, accumulator, count and filter on one dataset
# votes stores snack choices and prices stores the related numeric data.
# The sections below demonstrate indexing, accumulation, counting, and filtering.

votes = ["chips", "cookies", "chips", "fruit", "cookies", "chips",
         "pretzels", "cookies", "chips", "fruit", "chips", "cookies"]

prices = [2, 3, 2, 4, 3, 2, 1, 3, 2, 4, 2, 3]

# LESSON 1 - indexing
print(len(votes))
print(votes[-1])

# LESSON 2 - sum and average
total = 0
for price in prices:
    total += price

average = total / len(prices)
print("Total:", int(total))
print("Average:", int(average))

# LESSON 3 - count and filter
frequency = {}
for vote in votes:
    if vote in frequency:
        frequency[vote] += 1
    else:
        frequency[vote] = 1

print(frequency)

popular = []
for snack in frequency:
    if frequency[snack] >= 3:
        popular.append(snack)

print(popular)
```
