## Popcorn Hack 1: Read the API, Then Use It

The jersey range is changed to `1–55`, and the square-court example uses `225`, so the deterministic square-root result is `15.0`. The random game and jersey can vary on each run.

````python
# CODE_RUNNER: 3.14 Popcorn 1 - use math and random APIs

import math
import random

# API docstrings show what arguments these procedures accept and what they return.
print("random.randint:", random.randint.__doc__)
print("math.sqrt:", math.sqrt.__doc__)
print()

points_per_game = [24, 18, 31, 27, 22]

# random.choice can select any game, so this output is intentionally not predictable exactly.
print("Random game to review:", random.choice(points_per_game), "points")

# The changed range allows every whole-number jersey from 1 through 55, inclusive.
jersey = random.randint(1, 55)
print("Random jersey number (1-55):", jersey)

# A 225-square-foot square has side length 15 feet.
side = math.sqrt(225)
print("Square court side:", side, "feet")

# math.ceil rounds the calculated average upward to a whole number.
average = sum(points_per_game) / len(points_per_game)
print("Points per game:", average, "-> rounded up:", math.ceil(average))
````

## Popcorn Hack 2: Write a Flask Route

The roster is customized and a new `/coach` route follows the same Flask route pattern as `/score` and `/roster`.

````python
# CODE_RUNNER: 3.14 Popcorn 2 - customize Flask routes with a browser-safe fallback

# The published backend runner can use Flask, but the browser Pyodide fallback may not have it installed.
# This try/except keeps the real Flask version when available and uses the same route data locally when it is not.
try:
    from flask import Flask, jsonify

    app = Flask(__name__)

    # Each decorated procedure returns JSON when its matching address is requested.
    @app.route("/score")
    def score():
        return jsonify({"home": 6, "away": 4, "set": 3})

    @app.route("/roster")
    def roster():
        return jsonify({"players": ["Ruhaan", "Arya", "Deyar"]})

    # This is the extra route required by the Popcorn Hack.
    @app.route("/coach")
    def coach():
        return jsonify({"coach": "Coach Patel", "focus": "serve consistency"})

    # test_client calls the routes without opening a browser or separate server.
    client = app.test_client()
    print("/score says:", client.get("/score").get_json())
    print("/roster says:", client.get("/roster").get_json())
    print("/coach says:", client.get("/coach").get_json())

except ModuleNotFoundError:
    # Pyodide does not include Flask, so mirror the route responses with plain Python dictionaries.
    # The addresses and returned data stay the same, which still demonstrates the route/API pattern.
    routes = {
        "/score": {"home": 6, "away": 4, "set": 3},
        "/roster": {"players": ["Ruhaan", "Arya", "Deyar"]},
        "/coach": {"coach": "Coach Patel", "focus": "serve consistency"}
    }

    for route in ["/score", "/roster", "/coach"]:
        print(route, "says:", routes[route])
````

## Homework Hack: Tennis Season Stats Report

This version uses four standard-library modules: `json`, `statistics`, `datetime`, and `math`. The procedure documents its API and returns the required totals, average, best match, and points by opponent.

````python
# CODE_RUNNER: 3.14 Homework - tennis season stats report

import json
import math
import statistics
from datetime import date

# Each dictionary is one tennis match from the season report.
games = [
    {"date": "2026-09-05", "opponent": "Westview", "points": 12, "minutes": 64},
    {"date": "2026-09-12", "opponent": "Del Norte", "points": 15, "minutes": 71},
    {"date": "2026-09-19", "opponent": "Westview", "points": 10, "minutes": 58},
    {"date": "2026-09-26", "opponent": "Poway", "points": 18, "minutes": 83}
]

def summarize_season(season):
    """Return summary statistics for a list of tennis-match dictionaries.

    Parameters:
        season: list of dictionaries containing date, opponent, points, and minutes.
    Returns:
        dictionary with total_points, average_points, best_game, and points_by_opponent.
    """
    # statistics.mean and built-ins summarize the numeric results without rewriting tested library algorithms.
    points = [game["points"] for game in season]

    # Accumulate points under each opponent name so repeat opponents are combined.
    points_by_opponent = {}
    for game in season:
        opponent = game["opponent"]
        points_by_opponent[opponent] = points_by_opponent.get(opponent, 0) + game["points"]

    return {
        "total_points": sum(points),
        "average_points": round(statistics.mean(points), 1),
        "best_game": max(points),
        "points_by_opponent": points_by_opponent
    }

report = summarize_season(games)

# datetime subtracts dates; max keeps the report sensible even after the chosen date passes.
next_game = date(2026, 10, 15)
report["days_until_next_game"] = max(0, (next_game - date.today()).days)

# math is a fourth standard-library module; it rounds total playing time up to whole hours.
total_minutes = sum(game["minutes"] for game in games)
report["practice_hours_equivalent"] = math.ceil(total_minutes / 60)

# json.dumps converts the Python dictionary into readable JSON text for a web-style report.
print(json.dumps(report, indent=2))

# requirements.txt lists only external packages; standard-library modules already come with Python.
requirements_txt = """pandas
Flask
# json, math, statistics, and datetime are standard-library modules, so they are not listed.
"""
print("\nrequirements.txt:\n" + requirements_txt)
````
