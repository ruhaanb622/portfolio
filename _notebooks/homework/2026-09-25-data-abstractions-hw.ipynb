## Homework Hack: Grow the Level Collection

Each element in `level_stars` represents the stars collected in one level, stored in level order.

```python
# CODE_RUNNER: 3.2 Homework - update, append and traverse the level collection
# Store each level's stars in one ordered list instead of separate variables.
# Guard the second-level update so shorter lists do not cause an index error.
# Traverse the list to calculate the total without hardcoding each element.
level_stars = [12, 18, 10]

if len(level_stars) > 1:
    level_stars[1] = level_stars[1] + 5

level_stars.append(7)

total_stars = 0
for stars in level_stars:
    total_stars = total_stars + stars

print(level_stars)
print(len(level_stars))
print(total_stars)
```

## Tests

Main test should produce `[12, 23, 10, 7]`, `4`, and `52`. The empty-list test should produce `0` rounds and total `0`.

```python
# CODE_RUNNER: 3.2 Test - empty list total
# Use an empty list to verify the program works when there are no levels yet.
# The length check prevents invalid access to a missing second element.
# The loop runs zero times, so the total correctly stays at 0.
level_stars = []

if len(level_stars) > 1:
    level_stars[1] = level_stars[1] + 5

total_stars = 0
for stars in level_stars:
    total_stars = total_stars + stars

print(len(level_stars))
print(total_stars)
```

## Explanation

A list avoids creating a separate variable for every level and avoids rewriting the total calculation whenever another level is added. The same traversal works whether there are zero levels, four levels, or one hundred levels.

## Design Thinking

**Player/maker need:** A game maker needs to add and reorder levels without maintaining a separate variable for every level.

**Goal:** Keep all level star totals in one ordered collection that can grow without changing the traversal code.

**Representation considered:** Separate variables could store the same values, but a list is easier to extend and process as one collection.

**Prototype:** Store level scores in `level_stars`, update the second level, append a new level, then traverse the list for the total.

**Test/revision:** The main test produced four levels totaling 52. The empty-list test produced zero levels and a total of 0, so the second-level update is guarded by a length check. Clear level labels can also help players understand the levels without needing to know list indices.
