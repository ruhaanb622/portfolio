## Popcorn Hack 1: Translate Pseudocode to Python

The loop adds `3 + 5 + 2 + 4` for a total of `14`, then divides by four values for an average of `3.5`.

````python
# CODE_RUNNER: 3.09 Popcorn 1 - Translate pseudocode into Python
# Step 1: store the volunteer hours in a list.
hours = [3, 5, 2, 4]

# Step 2: start the running total at zero.
total = 0

# Step 3: visit each value and add it to the running total.
for h in hours:
    total = total + h

# Step 4: divide by the number of values to compute the average.
average = total / len(hours)

# Step 5: display the required results.
print(total)
print(average)
````

## Popcorn Hack 2: Predict, Then Fix

**Prediction before fixing:** `1`, because `count > capacity` counts only `15` and incorrectly excludes the boundary value `12`.  
**Fix:** change `>` to `>=`, which correctly counts both full workshops and prints `2`.

````python
# CODE_RUNNER: 3.09 Popcorn 2 - Fix the full-workshop counter
# The boundary matters: exactly 12 signed up is also full.
signed_up = [12, 7, 15, 9]
capacity = 12

# Count every workshop at or above capacity.
full = 0
for count in signed_up:
    if count >= capacity:
        full = full + 1

print("Full workshops:", full)
````

## Popcorn Hack 3: Modify a Counter into a Collector

Instead of only counting matches, this algorithm collects the names of workshops with fewer than 12 people signed up.

````python
# CODE_RUNNER: 3.09 Popcorn 3 - Modify the counter into a collector
workshops = ["Self-Defense", "Parenting Skills", "Fitness", "Self-Sufficiency"]
signed_up = [12, 7, 15, 9]
capacity = 12

# Start with an empty collector for workshop names.
open_workshops = []

# Compare each workshop's signup count with the capacity.
for i in range(len(workshops)):
    if signed_up[i] < capacity:
        open_workshops.append(workshops[i])

# Display both the number of matches and the collected names.
print("Workshops with open seats:", len(open_workshops))
print(open_workshops)
````

## Homework Hack: Supply Closet Restock Report

This combines three known algorithms: collect shortages, sum them with a loop, and find the maximum shortage. Items exactly at target are not included.

````python
# CODE_RUNNER: 3.09 Homework - Safe Passage Heals supply restock report

items = ["Toothbrushes", "Blankets", "Socks", "Shampoo", "Notebooks"]
in_stock = [40, 8, 25, 5, 30]
target = [30, 20, 25, 15, 30]

# Step 1: COLLECT every item below target and its shortage amount.
short_items = []
short_amounts = []
for i in range(len(items)):
    if in_stock[i] < target[i]:
        short_items.append(items[i])
        short_amounts.append(target[i] - in_stock[i])

# Step 2: SUM the shortages with a loop instead of sum().
total_needed = 0
for amount in short_amounts:
    total_needed = total_needed + amount

# Step 3: FIND THE MAX shortage while keeping its matching list position.
largest_index = 0
if len(short_amounts) > 0:
    for i in range(1, len(short_amounts)):
        if short_amounts[i] > short_amounts[largest_index]:
            largest_index = i

# Step 4: COMBINE the results into the final report, including the fully-stocked case.
if len(short_items) == 0:
    print("Fully stocked!")
else:
    print("Restock needed:")
    for i in range(len(short_items)):
        print(short_items[i] + ": need " + str(short_amounts[i]))
    print("Total items needed:", total_needed)
    print("Most urgent: " + short_items[largest_index] + " (short by " + str(short_amounts[largest_index]) + ")")
````
