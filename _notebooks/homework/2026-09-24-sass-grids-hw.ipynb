## Popcorn Hack 2: Size the Control Beaker

Four `fr` values place four beakers in one row; the last `2fr` gives the Control Beaker twice the width of any other beaker. This lesson explicitly allows a `grid-template-columns` style for the `fr` hack.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 2 — Control Beaker Twice as Wide | interactive: true -->
<div class="ocs__grid ocs__grid--standard" style="grid-template-columns: 1fr 1fr 1fr 2fr;">
  <div class="ocs__grid-cell">Beaker 1</div>
  <div class="ocs__grid-cell">Beaker 2</div>
  <div class="ocs__grid-cell">Beaker 3</div>
  <div class="ocs__grid-cell">Control Beaker</div>
</div>
```

## Homework Hack: Temperature Trial and Calculator

Click **Run** to check the editable code and preview both grids. The runner shows **PASS** or **FIX** for each lesson rule. The temperature title spans the standard grid, Beaker 3 is accented as the outlier, and the calculator's Avg key is accented.

```python
%%html
<!-- UI_RUNNER: Homework Hack — Temperature and Calculator Grids with Live Checks | interactive: true, render_mode: javascript -->
<script>
const myHTML = [
  '<div class="ocs__grid ocs__grid--standard">',
  '  <div class="ocs__grid-cell ocs__grid-cell--header">Temperature Trial (°C)</div>',
  '  <div class="ocs__grid-cell">Beaker 1: 22.4</div>',
  '  <div class="ocs__grid-cell">Beaker 2: 22.9</div>',
  '  <div class="ocs__grid-cell ocs__grid-cell--accent">Beaker 3: 31.6</div>',
  '</div>',
  '<br>',
  '<div class="ocs__grid ocs__grid--calculator">',
  '  <div class="ocs__grid-cell">7</div>',
  '  <div class="ocs__grid-cell">8</div>',
  '  <div class="ocs__grid-cell">9</div>',
  '  <div class="ocs__grid-cell ocs__grid-cell--accent">Avg</div>',
  '</div>'
].join('');

let classList = [];
for (const found of (myHTML.match(/class="[^"]*"/g) || [])) {
  classList = classList.concat(found.slice(7, -1).split(' ').filter(Boolean));
}
const madeUp = classList.filter(name => !name.startsWith('ocs__') && !name.startsWith('cols-'));
function line(label, ok) {
  return '<li>' + (ok ? 'PASS  ' : 'FIX  ') + label + '</li>';
}
const checks = '<ul>' +
  line('No inline style attributes', !myHTML.includes('style=')) +
  line('No made up classes' + (madeUp.length ? ' (found: ' + madeUp.join(', ') + ')' : ''), madeUp.length === 0) +
  line('Has an ocs__grid root', classList.includes('ocs__grid')) +
  line('Uses the standard variant for Grid 1', classList.includes('ocs__grid--standard')) +
  line('Uses the calculator variant for Grid 2', classList.includes('ocs__grid--calculator')) +
  line('Every box uses ocs__grid-cell', classList.filter(name => name === 'ocs__grid-cell').length >= 8) +
  line('Title is a header cell', classList.includes('ocs__grid-cell--header')) +
  line('Has accent cells (outlier beaker and Avg key)', classList.filter(name => name === 'ocs__grid-cell--accent').length >= 2) +
  '</ul>';
outputElement.innerHTML = checks + myHTML;
</script>
```
