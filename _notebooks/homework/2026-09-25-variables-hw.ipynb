## Homework Hack: Build the UESL Star Counter

The game uses a string name, a Boolean reduced-motion setting, and three separate numeric level-star variables. The starting total is `40`. After a bonus of `5` is added to level two, the stored total stays `40` until it is recalculated, when it becomes `45`.

```python
# CODE_RUNNER: UESL 3.1 Homework - variables, assignment, and recalculation
# Store the game name and accessibility setting using appropriate data types.
game_name = "Star Trail"
reduced_motion = True

# Each level starts with the values required by the lesson.
level_one_stars = 12
level_two_stars = 18
level_three_stars = 10

# Calculate and store the starting total once.
total_stars = level_one_stars + level_two_stars + level_three_stars
print("Game:", game_name)
print("Reduced motion:", reduced_motion)
print("Original total:", total_stars)

# Add the required bonus to level two. The old total does not update automatically.
bonus = 5
level_two_stars = level_two_stars + bonus
print("Stored total after bonus:", total_stars)

# Reassign total_stars to recalculate it from the updated level values.
total_stars = level_one_stars + level_two_stars + level_three_stars
print("Recalculated total:", total_stars)
```

## Tests

**Bonus 5 prediction/result:** original `40`, stored `40`, recalculated `45`.  
**Bonus 0 prediction/result:** original `40`, stored `40`, recalculated `40`.

```python
# CODE_RUNNER: UESL 3.1 Tests - verify both required bonus cases
# Reinitialize every test so one test cannot affect the next one.
for bonus in [5, 0]:
    level_one_stars = 12
    level_two_stars = 18
    level_three_stars = 10
    total_stars = level_one_stars + level_two_stars + level_three_stars
    original_total = total_stars
    level_two_stars = level_two_stars + bonus
    stored_total = total_stars
    total_stars = level_one_stars + level_two_stars + level_three_stars
    print("Bonus:", bonus, "| Original:", original_total, "| Stored:", stored_total, "| Recalculated:", total_stars)
```

## Design Thinking

**Player need:** Some players may be uncomfortable with fast movement or animation.  
**Goal:** Give the player control over reduced motion while keeping game score state easy for the maker to understand.  
**Representation considered:** A Boolean is appropriate for reduced motion because the setting has two states, on or off. Numeric variables are appropriate for star counts.  
**Prototype:** The star counter stores three level scores, a total, and the reduced-motion setting.  
**Test/revision:** Testing bonuses `5` and `0` showed that a stored total must be explicitly recalculated after a level value changes. Reduced motion should remain a player choice because accessibility needs differ between players.
