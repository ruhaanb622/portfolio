## Homework Hack: Interactive Settings Panel

This solution uses three labeled OCS toggles. Each toggle controls a real page feature, and the status panel updates the number of toggles currently turned on.

```python
%%html
<!-- UI_RUNNER: Homework Hack — Interactive Toggle Settings | interactive: true -->
<!-- Each label uses the OCS toggle classes so SASS handles the switch styling for us. -->
<!-- I use .checked in JavaScript because it directly tells whether a checkbox toggle is on or off. -->

<div class="ocs__container">
  <div class="ocs__card">
    <h2 class="ocs__section-title">My Website Settings</h2>
    <p class="ocs__description">Flip the switches to change features on this panel.</p>

    <label class="ocs__toggle ocs__toggle--green">
      <input id="notifications-toggle" class="ocs__toggle-input" type="checkbox">
      <span class="ocs__toggle-track" aria-hidden="true"></span>
      <span class="ocs__toggle-label">Show Notifications</span>
    </label>

    <label class="ocs__toggle ocs__toggle--green">
      <input id="focus-toggle" class="ocs__toggle-input" type="checkbox">
      <span class="ocs__toggle-track" aria-hidden="true"></span>
      <span class="ocs__toggle-label">Focus Mode</span>
    </label>

    <label class="ocs__toggle ocs__toggle--green">
      <input id="tips-toggle" class="ocs__toggle-input" type="checkbox">
      <span class="ocs__toggle-track" aria-hidden="true"></span>
      <span class="ocs__toggle-label">Show Study Tip</span>
    </label>

    <div class="ocs__callout">Toggles currently on: <output id="toggle-count" aria-live="polite">0</output></div>
    <p id="notification-message" hidden>🔔 Notifications are enabled.</p>
    <div id="focus-panel" class="ocs__card" hidden><strong>Focus Mode:</strong> Extra distractions are hidden so you can concentrate.</div>
    <p id="study-tip" hidden>💡 Study Tip: Break a large task into smaller steps.</p>
  </div>
</div>

<script>
  const toggles = document.querySelectorAll('.ocs__toggle-input');
  const countOutput = document.getElementById('toggle-count');
  const notificationMessage = document.getElementById('notification-message');
  const focusPanel = document.getElementById('focus-panel');
  const studyTip = document.getElementById('study-tip');

  // Recount all checked toggles whenever one changes so the displayed number stays accurate.
  function updateToggleCount() {
    let count = 0;
    toggles.forEach(function(toggle) {
      if (toggle.checked) count += 1;
    });
    countOutput.textContent = count;
  }

  // Each switch controls a real visible feature, not just the appearance of the toggle itself.
  document.getElementById('notifications-toggle').addEventListener('change', function(event) {
    notificationMessage.hidden = !event.target.checked;
    updateToggleCount();
  });
  document.getElementById('focus-toggle').addEventListener('change', function(event) {
    focusPanel.hidden = !event.target.checked;
    updateToggleCount();
  });
  document.getElementById('tips-toggle').addEventListener('change', function(event) {
    studyTip.hidden = !event.target.checked;
    updateToggleCount();
  });
</script>
```
