## Python 3.08 Iterations

Completed Popcorn and Homework Hacks using `for`, `continue`, `break`, counters, running totals, and thresholds.

## Popcorn Hack: Wildfire Dispatch Scanner

```python
# CODE_RUNNER: Popcorn Hack - wildfire dispatch scanner
# Scan each wildfire alert and classify it using the cutoff values.
# continue skips safe alerts; break stops the loop when a critical alert appears.
# Counters and a running total summarize the alerts processed.

alerts = [1, 4, 2, 6, 3, 9, 5]

safe_cutoff = 3
critical_cutoff = 9

action_count = 0
safe_count = 0
total_action_score = 0
critical_found = False

for alert in alerts:
    if alert < safe_cutoff:
        safe_count += 1
        print("Alert " + str(alert) + ": safe, skipping")
        continue

    action_count += 1
    total_action_score += alert
    print("Alert " + str(alert) + ": action needed")

    if alert >= critical_cutoff:
        critical_found = True
        print("Critical alert found - stopping scan")
        break

print("Safe alerts skipped: " + str(safe_count))
print("Action alerts: " + str(action_count))
print("Total action score: " + str(total_action_score))
print("Critical found: " + str(critical_found))
```

## Homework Hack: Server Load Monitor

This checker uses nine values, two thresholds, `continue`, `break`, and multiple summary values.

```python
# CODE_RUNNER: Homework Hack - themed loop-based checker
# Check each server load using safe and critical thresholds.
# continue skips safe loads after counting them; break stops at a critical load.
# Tracking variables build the final report for everything checked before stopping.

server_loads = [22, 47, 35, 68, 51, 76, 44, 93, 59]
safe_cutoff = 40
critical_cutoff = 90

checked_count = 0
action_count = 0
skipped_count = 0
total_score = 0
critical_found = False

for load in server_loads:
    checked_count += 1

    if load < safe_cutoff:
        skipped_count += 1
        print(str(load) + "% load: safe, no action")
        continue

    action_count += 1
    total_score += load
    print(str(load) + "% load: action needed")

    if load >= critical_cutoff:
        critical_found = True
        print("Critical server load found. Stop checking.")
        break

print("Final Report")
print("Checked: " + str(checked_count))
print("Skipped: " + str(skipped_count))
print("Actions: " + str(action_count))
print("Total load score: " + str(total_score))
print("Critical found: " + str(critical_found))
```
