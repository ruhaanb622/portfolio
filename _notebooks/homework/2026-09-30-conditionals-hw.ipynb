## Popcorn Hack 1: Fix the Silent Failure

With `amount_requested = 0`, the original code printed nothing. That is a logic bug, not a Python error, because the false outcome had no answer.

```python
# CODE_RUNNER: 3.6 Popcorn 1 - make both outcomes print

amount_requested = 0

# 1 condition -> 2 outcomes -> 2 answers
# The else prevents the false case from silently falling off the end.
if amount_requested > 100:
    print("This message is asking for a large amount of money.")
else:
    print("This message is not asking for a large amount of money.")
```

## Popcorn Hack 2: Fix the `elif` Order

In the broken version, `risk_score >= 6` and `risk_score >= 9` could never run because every score that satisfies them also satisfies the first `risk_score >= 3` branch. The fix is to test the highest threshold first.

```python
# CODE_RUNNER: 3.6 Popcorn 2 - put thresholds in reachable order

risk_score = 9

# Highest-to-lowest ordering keeps every branch reachable.
# 3 thresholds -> 4 outcomes -> 4 answers
if risk_score >= 9:
    print("Do not reply. This is almost certainly a scam.")
elif risk_score >= 6:
    print("This looks risky. Check with someone you trust before replying.")
elif risk_score >= 3:
    print("Be a little careful, but this is probably fine.")
else:
    print("Nothing suspicious in this message.")
```

## Homework Hack: Scam-Message Trainer

Each signal adds independently to the score, then a highest-to-lowest `if / elif / else` chain gives every possible score an answer.

```python
# CODE_RUNNER: 3.6 Homework - score a message, then advise on it

def score_and_advise(message, amount_requested, known_contact):
    # Start from zero so each scam signal can contribute its own points.
    risk_score = 0

    # No else is needed for these scoring checks because a missing signal should simply add zero points.
    if "gift card" in message.lower():
        risk_score += 4

    # This signal is independent, so it can add points even when another signal already matched.
    if amount_requested > 100:
        risk_score += 3

    # Again, no else is needed: a known contact contributes no extra risk points.
    if not known_contact:
        risk_score += 2

    # 3 thresholds -> 4 outcomes -> 4 answers
    # Highest thresholds come first so no more serious branch is hidden by an easier condition.
    if risk_score >= 7:
        advice = "Do not reply. This message has multiple strong scam signals."
    elif risk_score >= 4:
        advice = "This message looks risky. Verify it before responding."
    elif risk_score >= 1:
        advice = "Use caution and check the sender or request."
    else:
        advice = "No scam signals were detected by this trainer."

    print("Risk score:", risk_score)
    print(advice)
    return risk_score

# Test the exact starter message from the lesson: all three signals are present.
score_and_advise("URGENT: send gift cards now to unlock your account", 500, False)
```

## Reachability Proof

- **Score 9 / highest branch:** gift-card message, amount `500`, unknown contact → `4 + 3 + 2 = 9`.
- **Score 7 / risky branch:** gift-card message, amount `500`, known contact → `4 + 3 = 7`; to make the second branch distinct in testing below, use score `4` instead.
- **Score 4 / second branch:** gift-card message, amount `0`, known contact → `4`.
- **Score 2 / caution branch:** clean message, amount `0`, unknown contact → `2`.
- **Score 0 / safe branch:** clean message, amount `0`, known contact → `0`.

The separate scoring `if` statements are correct without `else` because each one asks whether to **add points**; doing nothing is the intended behavior when a signal is absent.

```python
# CODE_RUNNER: 3.6 Homework tests - prove every reporting branch is reachable

# Each case is chosen to hit a different reporting outcome, including both required extremes.
tests = [
    ("URGENT: send gift cards now", 500, False),   # score 9: all signals
    ("Please buy a gift card", 0, True),          # score 4: gift-card only
    ("Hi, checking in", 0, False),                # score 2: unknown contact only
    ("Hi Mom, see you Sunday", 0, True)           # score 0: completely clean
]

# Running every test confirms no input falls through without output.
for message, amount, known in tests:
    print("---")
    score_and_advise(message, amount, known)
```
