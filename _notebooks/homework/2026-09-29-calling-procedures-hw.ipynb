## Popcorn Hack 1: Simple Procedure Call and Parameters

Define `report_status(location, status)` and call it twice with different arguments.

```python
# CODE_RUNNER: Procedures Popcorn #1
# Parameters make this procedure reusable with different locations and statuses.
def report_status(location, status):
    print("Status update from " + location + ": " + status)

# These calls pass different arguments into the same procedure.
report_status("Poway", "Shelter open")
report_status("Rancho Bernardo", "Supplies delivered")
```

## Popcorn Hack 2: Decomposition and Helper Procedures

The helper `is_priority()` handles one small decision. The outer `process_report()` calls that helper and uses its returned Boolean to choose the response.

```python
# CODE_RUNNER: Popcorn Hack 2 - Decomposition & Helper Procedures
# The helper procedure returns a Boolean that the outer procedure can reuse.
def is_priority(incident_type):
    return incident_type == "Medical Emergency"

# The outer procedure delegates the priority check to the helper.
def process_report(location, incident_type):
    priority = is_priority(incident_type)
    if priority:
        print("PRIORITY: " + incident_type + " reported in " + location)
    else:
        print("STANDARD REPORT: " + incident_type + " reported in " + location)

# Test both a priority and a standard incident.
process_report("Poway", "Medical Emergency")
process_report("Poway", "Power Outage")
```

## Homework Hack: Emergency Response Dispatcher

This program uses a helper procedure named `check_priority()` and an outer procedure named `dispatch_response()`. The outer procedure calls the helper and prints a different response based on the Boolean it returns.

```python
# CODE_RUNNER: 3.12 Homework - emergency response dispatcher
# The helper identifies incidents that need immediate priority handling.
def check_priority(incident_type):
    return incident_type in ["Medical Emergency", "Fire"]

# The outer procedure calls the helper instead of repeating the priority logic.
def dispatch_response(location, incident_type):
    priority = check_priority(incident_type)
    if priority:
        print("PRIORITY DISPATCH: Send responders to " + location + " for " + incident_type)
    else:
        print("STANDARD DISPATCH: Record " + incident_type + " in " + location)

# Three calls use different incident types and more than one location.
dispatch_response("Poway", "Medical Emergency")
dispatch_response("Rancho Bernardo", "Power Outage")
dispatch_response("Poway", "Fire")
```

## Explanation

Using a helper procedure keeps the priority rule in one place instead of repeating the same comparison inside every dispatch. Parameters make both procedures reusable because the same code can process different locations and incident types.
