## Homework Hack 1: Key-Value

Create a user profile dictionary and add useful information that could be tracked by a computer system.

```python
# CODE_RUNNER: Homework Hack 1 - Key-Value
# A dictionary stores related user information as descriptive key-value pairs.
# items() lets the loop display every stored field without separate print statements.

user_profile = {
    "id": 101,
    "name": "Ruhaan",
    "email": "ruhaan@example.com",
    "active": True,
    "favorite_color": "blue",
    "login_count": 12,
    "last_activity": "Completed Variables homework"
}

print("User Profile:")
for key, value in user_profile.items():
    print(f"{key}: {value}")
```

## Homework Hack 2: Add-Dictionary

Add a dictionary to a user object so related profile information can be stored and updated together.

```python
# CODE_RUNNER: Homework Hack 2 - Add-Dictionary
# The User class groups related values into one reusable object.
# profile is a nested dictionary for extra information about the same user.
# Updating profile keys changes only those stored values.

class User:
    def __init__(self, user_id, name, email, active, scores, roles, last_login):
        self.id = user_id
        self.name = name
        self.email = email
        self.active = active
        self.scores = scores
        self.roles = roles
        self.last_login = last_login
        self.profile = {
            "favorite_color": "blue",
            "login_count": 12,
            "last_activity": "Variables lesson"
        }

    def __repr__(self):
        return f"User(id={self.id}, name={self.name}, email={self.email}, profile={self.profile})"

alice = User(101, "Alice", "alice@example.com", True, [0.91, 0.87, 0.76], ["student"], None)

print(alice)
print("Favorite color:", alice.profile["favorite_color"])

alice.profile["login_count"] += 1
alice.profile["last_activity"] = "Finished homework"

print("Updated profile:", alice.profile)
```

## Homework Hack 3: Add-Behavior

Add useful methods to the User class so an object can update its own data and perform actions.

```python
# CODE_RUNNER: Homework Hack 3 - Add-Behavior
# Methods give the User object behaviors for changing scores, roles, and status.
# average_score safely returns 0 when the score list is empty.
# The method calls at the end demonstrate how the object updates its own data.

class User:
    def __init__(self, user_id, name, email, active, scores, roles, last_login):
        self.id = user_id
        self.name = name
        self.email = email
        self.active = active
        self.scores = scores
        self.roles = roles
        self.last_login = last_login

    def __repr__(self):
        return (
            f"User(id={self.id}, name={self.name}, email={self.email}, "
            f"active={self.active}, scores={self.scores}, roles={self.roles}, "
            f"last_login={self.last_login})"
        )

    def add_score(self, score):
        self.scores.append(score)

    def average_score(self):
        return sum(self.scores) / len(self.scores) if self.scores else 0

    def add_role(self, role):
        if role not in self.roles:
            self.roles.append(role)

    def deactivate(self):
        self.active = False

alice = User(101, "Alice", "alice@example.com", True, [0.91, 0.87, 0.76, 0.55, 0.92], ["student", "scrummer"], None)

print(alice)
print("Average score:", round(alice.average_score(), 2))

alice.add_score(0.95)
alice.add_role("developer")
alice.deactivate()

print("Updated average:", round(alice.average_score(), 2))
print("Updated roles:", alice.roles)
print("Active:", alice.active)
print(alice)
```
